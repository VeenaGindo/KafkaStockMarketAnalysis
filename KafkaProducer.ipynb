```python
import pandas as pd
from kafka import KafkaProducer
from time import sleep
from json import dumps
import json
```

```python
producer = KafkaProducer(bootstrap_servers=['13.127.122.128:9092'], #change ip here
                         value_serializer=lambda x: 
                         dumps(x).encode('utf-8'))
```

```python
producer.send('open_stock', value={'name':'Sherly'})
```

```python
df = pd.read_csv("indexProcessed.csv")
```

```python
while True:
    dict_stock =df.sample(1).to_dict(orient="records")[0]
    producer.send("open_stock",value=dict_stock)
    sleep(1)
```
